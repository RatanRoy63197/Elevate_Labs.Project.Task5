```python
import pandas as pd
import matplotlib as mptb
import seaborn as snb
```

```python
titanic=pd.read_csv(r"C:\Users\Xervice5433\Downloads\tested.csv")
titanic.head(11)
```

```python
titanic.isnull().sum()
```

```python
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
```

```python
titanic['Cabin'] = titanic['Cabin'].fillna('Unknown')
```

```python
titanic.isnull().sum()
```

```python
titanic['Fare'] = titanic['Fare'].fillna(titanic['Fare'].median())
```

```python
titanic.isnull().sum()
```

```python
titanic.describe()
```

```python
titanic.value_counts()
```

```python
# Q1.Gender wise Ticket Sold?
import matplotlib.pyplot as plt
titanic['Sex'].value_counts().plot.pie(autopct="%1.1f%%")
plt.title('Percentage of Sex wise Ticket Sold',fontsize=16)
plt.show()
```

```python
#Q2. What is the overall survival rate?
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# Survival rate
survival_rate = titanic['Survived'].mean() * 100
print(f"Overall Survival Rate: {survival_rate:.2f}%")

sns.countplot(x='Survived', data=titanic)
plt.title("Survival Distribution")
plt.show()
```

```python
#Q3. Does Gender Impact Survival?
sns.countplot(x='Sex', hue='Survived', data=titanic)
plt.title("Survival by Gender")
plt.show()
```

```python
#Q4. Does Passenger Class Affect Survival?
sns.countplot(x='Pclass', hue='Survived', data=titanic)
plt.title("Survival by Passenger Class")
plt.show()
```

```python
#Q5. What is the Age Distribution?
sns.histplot(titanic['Age'], kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
#Q6. Does Age Influence Survival?
sns.boxplot(x='Survived', y='Age', data=titanic)
plt.title("Age vs Survival")
plt.show()
```

```python
#Q7. Fare Distribution & Outliers?
sns.boxplot(x=titanic['Fare'])
plt.title("Fare Distribution & Outliers")
plt.show()
```

```python
#Q8. Correlation Analysis?
corr = titanic.corr(numeric_only=True)

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
#Q9. Pairwise Relationship Exploration?
sns.pairplot(titanic[['Age','Fare','Pclass','Survived']], hue='Survived')
plt.show()
```

```python
#Q10. Does Family Size Impact Survival?
titanic['FamilySize'] = titanic['SibSp'] + titanic['Parch']

sns.boxplot(x='Survived', y='FamilySize', data=titanic)
plt.title("Family Size vs Survival")
plt.show()
```

```python
#Q11. Embarkation Port Impact?
sns.countplot(x='Embarked', hue='Survived', data=titanic)
plt.title("Survival by Embarkation Port")
plt.show()
```

```python
#Q12. Statistical Skewness Check?
print(titanic.skew(numeric_only=True))
```

```python
#Q17. Correlation Strength Analysis?
titanic.corr(numeric_only=True)['Survived'].sort_values(ascending=False)
```

```python
#Q18. Family Size Effect Analysis?
titanic['FamilySize'] = titanic['SibSp'] + titanic['Parch']
titanic.groupby('FamilySize')['Survived'].mean()
```

```python
#Q19. Multivariable Interaction Analysis?
pd.crosstab(titanic['Pclass'], titanic['Sex'], values=titanic['Survived'], aggfunc='mean')
```
